```python
import pandas as pd 
students = pd.DataFrame({
    "Student_ID": ["S101", "S102", "S103", "S104", "S105",
                   "S106", "S107", "S108", "S108"],
    "Name": ["Anu", "Binu", "Charu", "Deepa", "Ebin",
             "Fathima", "Gokul", "Hana", "Hana"],
    "Course": ["BCA", "BSc", "BCA", "BSc", "BCA",
               "BSc", "BCA", "BSc", "BSc"],
    "Marks": [86, 72, 94, 65, 78, 91, None, 88, 88],
    "Attendance": [92, 78, 96, 70, 85, 93, 81, None, None],
    "City": ["Kozhikode", "Kannur", "Kochi", "Kozhikode",
             "Kannur", "Kochi", "Kozhikode", None, None],
    "Join_Date": ["2026-01-10", "2026-01-12", "2026-02-02",
                  "2026-02-15", "2026-03-01", "2026-03-09",
                  "2026-03-12", "2026-04-05", "2026-04-05"]
})

fees = pd.DataFrame({
    "Student_ID": ["S101", "S102", "S103", "S104",
                   "S105", "S106", "S107", "S109"],
    "Fee_Amount": [20000, 18000, 20000, 18000,
                   20000, 18000, 20000, 20000],
    "Amount_Paid": [20000, 10000, 20000, 0,
                    15000, 18000, 0, 20000],
    "Payment_Status": ["Paid", "Partial", "Paid", "Unpaid",
                       "Partial", "Paid", "Unpaid", "Paid"]
})

display(students)
display(fees)
```

B. Filtering, updating, sorting and statistics
---

7. Find students with Marks greater than 80.

```python
display(students[students["Marks"]>80])
```

8. Find BCA students whose Attendance is at least 85.

```python
display(students[(students["Course"] == "BCA")&(students["Attendance"]>=85)])
```

9. Find students from either Kozhikode or Kannur who scored below 80.

```python
display(students[((students["City"] == "Kozhikode") | (students["City"] == "Kannur")) & (students["Marks"] < 80)])
```

10. Add a Bonus_Marks column with 5 marks for every student and calculate Final_Marks.

```python
students["Bonus_Marks"] = 5
students["Final_Marks"] = students["Marks"] + students["Bonus_Marks"]
display(students)
```

11. Create a Result column: Pass for Marks of at least 70, Fail otherwise; keep missing Marks unclassified.

```python
students["Result"] = students["Marks"].apply(lambda x: "Pass" if x >=70 else "Unclassified" if pd.isna(x) else "Fail")
display(students)
```

12. Sort students by Marks from highest to lowest, placing missing values last.

```python
display(students.sort_values("Marks",ascending=False,na_position="last"))
```

17. Count missing values in each column using isna().

```python
print("Missing values:\n",students.isna().sum())
```

18. Fill missing Marks with the mean Marks and missing City with "Unknown". Explain the choice in a code comment.

```python
#Fill the missing marks with mean
students["Marks"] = students["Marks"].fillna(students["Marks"].mean())
#Fill missing City values with "Unknown"
students["City"] = students["City"].fillna("Unknown")
display(students)
```

19. Create a separate DataFrame that removes rows with missing Attendance using dropna().

```python
missing = students.dropna(subset=["Attendance"])
display(missing)
```

24. Merge students and fees using Student_ID with a left join. Which students have no matching fee record?

```python
a = students.merge(fees,on="Student_ID",how="left")
display(a)
display(a[a["Fee_Amount"].isna()]) 
```

25. Merge the two tables with an outer join. Which Student_ID exists only in Fee_Records?

```python
merged = students.merge(fees,how="outer")
display(merged)
display( merged[merged["Name"].isna()])
```

26. Split Student_Records into the first four rows and the remaining rows; combine them again vertically using concat() and ignore_index=True.

```python
first = students.iloc[:4]
second = students.iloc[4:]
display(first,second)
display(pd.concat([first,second],ignore_index=True))
```

31. Convert Join_Date to datetime and create Month, Year and Day_Name columns.

```python
students["Join_Date"] = pd.to_datetime(students["Join_Date"])
students["Month"] = students["Join_Date"].dt.month_name()
students["Year"] = students["Join_Date"].dt.year
students["Day_Name"] = students["Join_Date"].dt.day_name()
display(students)

```
