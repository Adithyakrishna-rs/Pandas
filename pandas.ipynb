Data frame : creating data
---

```python
df = pd.DataFrame({"Name":["Adithya","Alida","Midhuna","Amna","Riya"],"Age":[20,21,22,23,24],"Course":["Data Science","Full Stack","Data Analytics","Agentic AI","Data Science"],"Attendance":[50,98,70,95,90],"Mark":[50,98,99,65,76]})
display(df)
```

Inspecting data

```python
display(df.head(3))#print the columns you needed
display(df.tail(2)) #print last five rows
```

```python
print("Shapes:",df.shape)
print("Columns:",df.columns)
df.info()
display(df.describe())
```

Selecting Columns
---

```python
print(df["Name"])
display(df[["Name","Age"]])
display(df[["Age","Course","Attendance"]])
```

Selecting rows:loc and iloc
---

```python
print(df.loc[0,"Name"])
display(df.loc[1:3,["Name","Course"]])
display(df.iloc[0:3,0:2])

```

```python
display(df.loc[1:4,["Course","Attendance"]])
display(df.iloc[1:5,2:4])
```

Filtering rows
---

```python
display(df[df["Attendance"]>=75])
display(df[(df["Course"] == "Data Science") & (df["Attendance"] >= 85)])
```

```python
display(df[(df["Attendance"]>=90) | (df["Mark"]>=90)])
```

Adding and Updating columns
---

```python
df["Bonus"] = 5
df["Final_Marks"]= df["Mark"]+df["Bonus"]
def result(Final_Marks):
    if Final_Marks>= 75:
        return "Pass"
    else:
        return "Improve"
df["Result"] = df["Final_Marks"].apply(result)
df["Grade"] = df["Final_Marks"].apply(lambda x: "A" if x>=90 else "B" if x>=70 else "C" if x>=60 else "D")
# df["Result"] = df["Final_Marks"].apply(lambda x: "Pass" if x >=75 else "Improve")
display(df)
```

Sorting
---

```python
display(df.sort_values("Mark"))
display(df.sort_values("Mark",ascending=False))
```

Statistics and aggregations
---

```python
print("Total:",df["Mark"].sum())
print("Mean:",df["Mark"].mean())
print("Median:",df["Mark"].median())
print("Min/Max:",df["Mark"].min(),df["Mark"].max())
print("Standard Deviation:",df["Mark"].std())
print("Count:",df["Mark"].count())
print(df["Course"].value_counts())

```
